# ATIVIDADE 12 - Tabela de frequências e gráficos para variáveis qualitativas.

**Objetivo:** Reforçar a construção de tabelas de frequência absoluta, relativa e acumulada.

---

Uma pesquisa foi feita com 50 estudantes sobre o principal meio de transporte usado para ir até a escola. As respostas obtidas foram:

| Meio de transporte | Quantidade de estudantes |
| ------------------ | ------------------------ |
| Carro              | 12                       |
| Ônibus             | 20                       |
| Bicicleta          | 8                        |
| A pé               | 10                       |


Questão 1 - Monte uma tabela de frequências (Frequência Relativa e Frequência Acumulada)

```python
import pandas as pd

# Utilize o DataFrame
df = pd.DataFrame({
    'Meio de Transporte': ['Carro', 'Ônibus', 'Bicicleta', 'A Pé'],
    'Quantidade de Estudantes': [12, 20, 8, 10],
})

total = df['Quantidade de Estudantes'].sum()

df['Frequência Relativa'] = df['Quantidade de Estudantes'] / total
df['Frequência Acumulada'] = df['Frequência Relativa'].cumsum()

print(df)

```

Questão 2 - Crie um gráfico de barras

```python
import matplotlib.pyplot as plt

df.plot(x='Meio de Transporte', y='Quantidade de Estudantes', kind='bar')
plt.show()
```

Questão 3 - Crie um gráfico de pizza

```python
plt.pie(df['Quantidade de Estudantes'], labels=df['Meio de Transporte'], autopct='%1.1f%%')
plt.show()
```

Questão 4 - Interprete os resultados
- Qual meio de transporte é mais utilizado?
- Qual meio de transporte é menos utilizado?
- Qual é a diferença percentual entre o mais e o menos frequente?

```python
meioMaisFrequente = df.loc[df['Quantidade de Estudantes'].idxmax(), 'Meio de Transporte']
meioMenosFrequente = df.loc[df['Quantidade de Estudantes'].idxmin(), 'Meio de Transporte']
print(f". O meio de transporte mais frequente é: {meioMaisFrequente}")
print(f". O meio de transporte menos frequente é: {meioMenosFrequente}")

df['Porcentagem'] = (df['Quantidade de Estudantes'] / df['Quantidade de Estudantes'].sum()) * 100

difPontos = df.loc[df['Porcentagem'].idxmax(), 'Porcentagem'] - df.loc[df['Porcentagem'].idxmin(), 'Porcentagem']
print(f". A diferença é de {difPontos:.2f} %")
```
