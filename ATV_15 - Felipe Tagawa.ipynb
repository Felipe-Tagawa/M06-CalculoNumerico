## 1 -Teste dos Sinais (Amostras Pareadas)

Um fisioterapeuta quer saber se um novo exercício respiratório reduz a frequência respiratória dos pacientes.
Foram medidos os valores antes e depois em 8 pessoas:

| **Paciente** | **Antes** | **Depois** |
| ------------ | --------- | ---------- |
| 1            | 22        | 20         |
| 2            | 24        | 23         |
| 3            | 19        | 18         |
| 4            | 26        | 25         |
| 5            | 21        | 22         |
| 6            | 23        | 21         |
| 7            | 25        | 25         |
| 8            | 20        | 19         |

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Declare a tabela como numpy

tab = np.array([[22, 20], [24, 23], [19, 18], [26, 25], [21, 22], [23, 21], [25, 25], [20, 19]])
```

##### 1.1 - Quais são as hipóteses do Teste dos Sinais?

**Resposta:**

H₀ (Hipótese nula):

O novo exercício respiratório não altera a frequência respiratória dos pacientes.
Ou seja, qualquer diferença observada entre antes e depois é atribuída ao acaso.

H₁ (Hipótese alternativa):

O novo exercício respiratório reduz a frequência respiratória dos pacientes.
Isto é, espera-se que depois do exercício a frequência esteja menor.

##### 1.2 - Identifique diferenças positivas, negativas e empates

```python
# Calcula Diferenças: Depois - Antes
diff = tab[:,1] - tab[:,0]

positivas = np.sum(diff > 0)
negativas = np.sum(diff < 0)
empates = np.sum(diff == 0)

print("Diferenças:", diff)
print("Positivas:", positivas)
print("Negativas:", negativas)
print("Empates:", empates)
```

##### 1.3 - Gere um gráfico de barras mostrando as diferenças

```python
plt.bar(range(1, len(diff)+1), diff)
plt.xlabel("Paciente")
plt.ylabel("Diferença")
plt.title("Diferenças na Frequência Respiratória")
plt.axhline(0, linestyle='-')
plt.show()
```

##### 1.4 - Calcule o p-valor

```python
resultado = stats.binomtest(6, 7, 0.5, alternative='greater') # 6 ;e o número de diff negativas, 7 é o total de diff - empates e 0.5 é padrão (neg ou pos)

print(resultado)

print(f"O valor de pValor: {resultado.pvalue}")

if resultado.pvalue < 0.05:
    print("Resultado: REJEITA H₀ (p < 0.05)")
    print("Há evidência de diferença significativa entre antes e depois")
else:
    print("Resultado: ACEITA H₀ (p ≥ 0.05)")
    print("Não há evidência suficiente de diferença significativa")
```

##### 1.5 - Conclusão

**Resposta:** Analisando os resultados, podemos observar que não há evidência significativa de que os exercícios respiratórios sejam eficientes, ao menos nesse intervalo (Antes e Depois), ao nível de significância de 6.25%.
